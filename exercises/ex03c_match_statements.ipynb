## Simple Matching

Write a program that asks the user to enter "red", "green", or "blue".
Use match to print a message like "You chose red!", "You chose green!", or "You chose blue!".
If the input doesnâ€™t match, print "Unknown color".


```python
color_chose = input('Chose color: red, green or blue').lower()
match color_chose:
    case 'red'|'blue'|'green' as id:
        print(f'You chose {id}!')
    case _ : print('Unknown color')

```

## Matching Multiple Values

Ask the user to enter a day of the week.
Use match to print "Weekend" if the day is Saturday or Sunday, otherwise print "Weekday".
(Hint: lowercase the input first with .lower().)

```python
inter_day_of_week = input('Chose a day of the week: ').capitalize()

match inter_day_of_week:
    case 'Monday'| 'Tuesday'| 'Wednesday'| 'Thursday'| 'Friday' as weekday:
        print(f'Today is {weekday}')
    case 'Saturday'| 'Sunday':
        print('It a weekend')
    case _: print('It is not a week day')
```

## Variable Binding

Given a tuple (x, y), use match to check:

(0, 0) â†’ "Origin"

(x, 0) â†’ "On the X-axis"

(0, y) â†’ "On the Y-axis"

(x, y) â†’ "Point at x=..., y=..."

Try it with different values.


```python
coordinates = (int(input('chose coordinate on x: ')), int(input('chose coordinate on y: ')))

match coordinates:
    case (0, 0): print('Origin')
    case (_, 0): print('On the X-axis')
    case (0, _): print('On the Y-axis')
    case (_, _): print(f'Point at x={coordinates[0]}, y={coordinates[1]}')
```

## Guards (with if)

Write a program that asks the user for an integer.
Use match to check:

If the number is negative

If the number is even

Otherwise, treat it as odd

```python
integer_num = int(input('Chose a whole number: '))

match integer_num:
    case integer_num if integer_num <0 and integer_num % 2 == 0:
        print(f'The number is even')
    case integer_num if integer_num > 0 and integer_num % 2 == 0:
            print(f'The number is even')
    case _: print('Is odd')


```

## Combining Patterns

Write a program that asks the user to enter a number between 1 and 10.
Use match to:

Print "One" if the number is 1

Print "Two or Three" if the number is 2 or 3

Print "Greater than 5" if the number is > 5

Print "Other number" otherwise

```python
number = int(input('chose number from 1 to 10: '))

match number:
    case 1: print('One')
    case 2|3: print('Two or Three')
    case number if number > 5: print('Greater than 5')
    case _: print('Other number')
```

## Matching Sequences

Given a list of names, use match to:

Print "Two names: ..." if the list has exactly two items

Print "Three names: ..." if the list has exactly three items

Print "Many names: ..." if the list has more than three items

Test with:

["Alice", "Bob"]
["Alice", "Bob", "Charlie"]
["Alice", "Bob", "Charlie", "David"]



```python
on = True 
name_list = []
while on:
    name = input('Put a name: or put \'stop\' if you dont have a name ').lower()
    if name != 'stop': name_list.append(name)
    else:
        on = False

match name_list:
    case name_list if len(name_list) == 2:
        print(f'Two names: {name_list}')
    case name_list if len(name_list) == 3:
         print(f'Three names: {name_list}')
    case name_list if name_list > 3: 
        print(f'Many names: {name_list}')
```

7. Matching Dictionaries

Write a program with a dictionary representing a student, e.g.:

student = {"name": "Sara", "id": 123, "program": "Computer Science"}


Use match to:

Print the name and ID if both exist.

Print the name and program if both exist.

Print "Unknown student data" otherwise.

```python
student = {
    "name": "Sara",
    "id": 123,
    "program": "Computer Science"
    }
student_id = False
program = False

for i in student:
    if i == 'id':
        student_id = True 
    if i == 'program':
        program = True


match student:
    case student if student_id:
        print(f'Student name: {student['name']} and id: {student['id']}')
    case student if program:
        print(f'Student name: {student['name']} and program: {student['program']}')


```
