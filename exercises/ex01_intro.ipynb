#### `type()` - Check Data Types

```python
# Use type() to check the data type of each of the following variables:
# 
x = 42
y = 3.14
z = "Python"
is_active = True
print(type(x), type(y), type(z), type(is_active))
```

```python
# Use id() to check whether two variables pointing to the same value actually point to the same object:
# 
a = 100
b = 100
print(id(a),
       id(b))
```

### Challenge Activity
#### Part 4 - Discount Calculator

```python
# Part 4 – Challenge
# 
# Write a short script that:
# 
# Stores the discount of an item (e.g., 10%)
discount = 10
# Prompt the user to enter the price of an item and stores it in a variable
price = round(float(input('What is the product\s price? ')),2)
# Calculates the final cost after discount
final_cost = round(discount/100 * price, 2)
# Prints it nicely formatted
print(f'Your final price after discount of {discount}% is ${final_cost}')
```
