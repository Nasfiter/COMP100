Given:

-   Find the maximum

-   Find the sum

```python
nums = [3, 7, 2, 9, 5]
sum_n = sum(nums)
print(sum_n)
max_n = max(nums)
print(max_n)
min_n = min(nums)
print(min_n)
print(sorted(nums))

max_1 = nums[0]
for i in nums:
    if i > max_1:
        max_1 = i

print(max_1)

```

Write a program that:

Uses zip() to pair students with their grades

Iterates through the pairs and prints:

> Alice scored 85  
> Bob scored 92  
> Charlie scored 78  


Finds the highest score using an aggregation function.

```python
names = 'Alice scored', 'Bob scored', 'Charlie scored'
grades = 85, 92, 78

result_score = zip(names, grades)

for i in list(result_score):
    for n in i:
        print(f'{n}',end=' ')
    print()


```
