#### 2
Make a set from the string 'Mississippi'. 

Print the resulting set and a count of them.


```python
text = 'Mississippi'
text_to_set = set(text)
print(text_to_set)
print(len(text_to_set))
```

#### 3
Prompt the user for a word and then print the number of unique vowels in it.

```python
vowels = ['a', 'e', 'i', 'o', 'u']
word = input('Ant word: ')
word_set = set(word)

count = 0
for i in word_set:
    if i in vowels: count += 1

print(f'unique vowels in {word} is: {count}')
```

```python
a = {1, 2, 5, 5, 7, 0}
b = {3, 2, 5, 6, 10, 0}

union_set = a | b
print(union_set)
```

#### 9
Create two sets of letters from the words "python" and "typhoon".

1.  Find letters they have in common.
1.  Find letters unique to "python".
1.  Find letters unique to "typhoon".
1.  Find all unique letters from both words. *[Use the results of the previous two steps]*

```python
a = set('python')
b = set('typhoon')
common = a & b
print(common)
unique_a = a - b # empty set 
print(unique_a)
unique_b = b - a  # empty set 
print(unique_b)

unique = unique_a | unique_b
print(unique)

```
