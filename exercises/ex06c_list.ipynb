### sort()

#### 11
Sort the list [3, 1, 4, 1, 5, 9].

Sort the list ['pear', 'apple', 'banana'].

Sort the same list by length of the words (use key=len).

```python
list_num = [3, 1, 4, 1, 5, 9]
list_num.sort()
print(list_num)

fruets = ['pear', 'apple', 'banana']
fruets.sort()
print(fruets)

fruets.sort(key=len, reverse=False)
print(fruets)
```

### Copy vs Assignment

#### 12
Create a list a = [1, 2, 3]. Assign it to b = a.

Append 4 to a. What happens to b?

Create a shallow copy with c = a.copy(). Append 5 to c. Does a change?

```python
a = [1, 2, 3]
b = a 
print(b)
a.append(4)
print(b)

c = a.copy()
print(a)
print(c.append(5))
```

### Shallow vs Deep Copy

#### 13
1.  Create nested = [[1, 2], [3, 4]].

    -   Print both nested and copy1.

1.  Repeat with deepcopy from the copy module. Whatâ€™s different?

```python
import copy
nested = [[1, 2], [3, 4]]
copy_q = nested.copy()
copy_q[0].append(99)
print(nested)
print(copy_q)
print(id(nested))
print(id(copy_q))



nested_1 = [[1, 2], [3, 4]]
copy_2 = copy.deepcopy(nested_1)
copy_2.append(99)

print(nested_1)
print(copy_2)



```
