#### 1
Create a dictionary that maps the first 5 positive integers to their squares.

##### Sample output
```python
{1: 1, 2: 4, ...}
```

````python
result = {key: key ** 2 for key in range(1, 6)}
print(result)
````


````python
name_age = {}

while True:
	name = input('Name: ')
	age = input('Age: ')
	name_age[name] = age
	if input('Do you want to continue? (yes/no)').lower() == 'no':
		break

print(name_age)
````
