Use a dictionary to simulate a phonebook (name â†’ phone number). Then:

1.  add a new contact
1.  update an existing contact

````python
phone_book = {}

def add_contact(data):
    data[input('Input name: ')] = input('Number: ')

def update_contact(name, data):

    if name in data:
        def condition():
            condition = int(input('To update name press: 1 \nTo update number press: 2 '))
            if condition == 1:
                number = data[name]
                del data[name]
                data[input('New name: ')] =  number
            elif condition == 2:
                data[name] = input('Input number: ')
            else:
                print('Invalid intrance')
                condition()
        condition()
    else:
        print(f'Contact {name} has not found.')

def delete_contact(name, data):
    del data[name]




while True:
    todo = int(input('If you want to add contact press: 1 \nIf. you want to update contact press: 2\nIf you want to delete contact press 3:\nTo exit press 4: '))
    if todo == 1:
        add_contact(data=phone_book)
        print(phone_book)
    elif todo == 2:
        print(phone_book)
        name_to_update = input('Name to update: ')
        if name_to_update in phone_book:
            update_contact(name=name_to_update, data=phone_book)
            print(phone_book)
        else:
            print('Contact number has not found: ')
    elif todo == 3:
        name_to_delete = input('Input contact name: ')
        if name_to_delete in phone_book:
            delete_contact(name=name_to_delete, data=phone_book)
        else:
            print(f'Contact with name {name_to_delete} has not found')
        print(phone_book)
    else:
        if todo != 1 or todo != 2 or todo != 3:
            break


````

#### 6
Given a dictionary of student names and their marks, write code to:

1.  print the highest score
1.  print the name of the student with the lowest score

````python
grades = {
    'Sasha': 90,
    'Alex': 100,
    'Sim': 10,
    'Dima': 85,
}

highest_score = 0 
best_student = None
lowest_acore = 100
worst_student = None
for key, valu in grades.items():
    if valu > highest_score:
        highest_score = valu
        best_student = key
    if valu < lowest_acore:
        lowest_acore = valu
        worst_student = key

print(f'Highest score got a student: {best_student}: {highest_score}')
print(f'Lowest score got a student {worst_student}: {lowest_acore}')
````

##### Frequency Counter
Write a program that counts the frequency of each character in a string and stores the result in a dictionary.

````python
word = 'ninqwifnoipqwnfpoiewnfoienwr'

frequency_count = {}

for i in word:
    if i in frequency_count:
        frequency_count[i] += 1
    else:
        frequency_count[i] = 1
print(frequency_count)
````

#### 8
Write the statements that will create a dictionary that maps numbers from 1 to 10 to either 'even' or 'odd'.

##### Sample output
`{1: 'odd', 2: 'even', 3: 'odd', 4: 'even', 5: 'odd', 6: 'even', 7: 'odd', 8: 'even', 9: 'odd', 10: 'even'}`

````python

dictionary = {}

for i in range(1, 11):
    if i % 2 == 0:
        dictionary[i] = 'even'
    else:
        dictionary[i] = 'odd'
print(dictionary)
````

#### 10
Given a dictionary of items and their prices, increase each price by 10%.  
*Hint: You may use the `round(value, precision)` function to make the results more pleasing*

````python
products = {
    'iphone': 1000,
    'ipad': 800,
    'macbook': 2000,
}

for i in products:
    products[i] = round(products[i] * 1.1)

print(products)
````

#### 11
##### Merge Two Dictionary
Write code to merge two dictionaries into one. If a key exists in both, add their values.

````python
dict1 = {
    "apple": 5,
    "banana": 3,
    "orange": 7,
    'mac': 9,
    'iphon': 10,
}

dict2 = {
    "banana": 4,
    "orange": 2,
    "grape": 6
}

for i in dict2:
    if i in dict1:
        dict1[i] += dict2[i]
    else:
        dict1[i] = dict2[i]

print(dict1)
````
