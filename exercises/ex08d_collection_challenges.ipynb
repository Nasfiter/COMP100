#### 11: Filter Dictionary
Keep only items from the dictionary whose values are even numbers.

with data `{'a': 3, 'b': 10, 'c': 7, 'd': 2}`

#### 14
Given the following dictionary 

```python
{'Ilia': 45, 'Arben': 43, 'Narendra': 46, 'Hao': 41, 'Mayy': 44, 'Yin': 42, 'Mehrdad': 44}
```

1. print the highest score
1. the name that corresponds with the lowest score

##### Sample output
```python
46
Narendra
```


````python
dictionary = {'Ilia': 45, 'Arben': 43, 'Narendra': 46, 'Hao': 41, 'Mayy': 44, 'Yin': 42, 'Mehrdad': 44}

highest_score = 0
lowest_score = 100
worst_student = None

for key, value in dictionary.items():
    if value > highest_score:
        highest_score = value
    if value < lowest_score:
        lowest_score = value
        worst_student = key

print(highest_score)
print(worst_student)
````

#### 15: Merge Dictionaries
Merge the two dictionaries below. If a key appears in both, sum their values.

##### Sample output:
with data {'x': 3, 'y': 7, 'z': 1} and {'y': 5, 'z': 2, 'w': 9}' }
```python
{'x': 3, 'w': 9, 'z': 3, 'y': 12}
```

````python
dict_1 = {'x': 3, 'y': 7, 'z': 1}
dict_2 = {'y': 5, 'z': 2, 'w': 9}

for i in dict_2:
    if i in dict_1:
        dict_1[i] += dict_2[i]
    else:
        dict_1[i] = dict_2[i]

print(dict_1)

````

#### 16: Count Word Frequency
Given the list below, create a dictionary that stores each word and how many times it appears.

##### Sample output:
with data `['apple', 'banana', 'apple', 'orange', 'banana', 'apple']`
```python
{'banana': 2, 'apple': 3, 'orange': 1}
```

````python
list_input = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
output = {}

for i in list_input:
    if i in output:
        output[i] += 1
    else:
        output[i] = 1

print(output)
````

#### 17: Reverse Dictionary Lookup
You are given a dictionary that maps courses to room numbers. Write code to invert it so rooms map to courses.

##### Sample output:
with data `{  'Python': 'B210', 'Java': 'B105', 'C#': 'B330' }`
```python
{'B210': 'Python', 'B105': 'Java', 'B330': 'C#'}
```

#### 19
Write a dictionary comprehension that maps numbers from 1 to 10 to either 'even' or 'odd'.

##### Sample output
```python
{1: 'odd', 2: 'even', 3: 'odd', 4: 'even', 5: 'odd', 6: 'even', 7: 'odd', 8: 'even', 9: 'odd', 10: 'even'}
```

````python

output = {}

for i in range(1, 11):
    if i % 2 == 0:
        output[i] = 'even'
    else:
        output[i] = 'odd'

print(output)
````
