#### 10 - Find Minimum

Write a function find_min(numbers) that returns the smallest number from a list.

Your output should look like:

The minimum of [3, 1, 4, 1, 5, 9, 2, 6, 5] is 1

The minimum of [-10, -20, 0, 10, 20] is -20

```python
def min_number(list_n_input:list[float|int]) ->float|int:
    '''
    Find the minimum number in the list
    Args:
      list_n_input(list[int|float]): list of numbers 
    Returns:
    float: The smallest number in the list.
    '''
    min_number = list_n_input[0]
    for n in list_n_input:
        if n < min_number:
            min_number = n
    return f'The minimum of {list_n_input} is {min_number}'

min_n_1 = min_number([3, 1, 4, 1, 5, 9, 2, 6, 5])
print(min_n_1)
min_n_2 = min_number([-10, -20, 0, 10, 20]) 
print(min_n_2)
```

The average of [10, 20, 30, 40, 50] is 30.0

```python
def avarage_number(numbers_list:list[int|float]) ->float:
    '''
    find an avarage number
    Arg: list of numbers
    Returns: avarage number -> float
    '''
    sum = 0 
    for n in numbers_list:
        sum += n
    return f'The average of {numbers_list} is {sum/len(numbers_list)}'

print(avarage_number([1, 2, 3]))
print(avarage_number([10, 20, 30, 40, 50]))
print(avarage_number([5, 15, 25, 35, 45, 55]))
```

```python
def squares(list_num:list[int|float]) ->list:
    '''finding square from the list input
        Arg: list of numbers
        Returns: liat of square numbers 
    '''
    result = []
    for n in list_num:
        aq_n = n ** 2
        result.append(aq_n)
    return f'The squares of {list_num} are {result}'

print(squares(list_num=[1, 2, 3, 4, 5]))
print(squares(list_num=[6, -7, 1.2]))
```

#### 14 - Ticket Price Calculator

Write a function ticket_price(age, discount=0.1) that calculates ticket price.

Base price = $20. Apply discount only if discount is given.

```python
def start():

    BASE_PRICE = 20
    age = int(input('Age: '))

    if age < 4 or age > 100:
        print('Invalid age')
        start()

    def ticket_price(buyer_age:int, price_t:int|float, discount:float=0.1) ->float:
        if  buyer_age <= 18 or buyer_age >= 60:
            price = price_t * (1 - discount)
            return f'Your ticket price after discount: ${price:.2f}'
        else:
            return f'Your ticket price: ${price_t:.2f}'

    final_price = ticket_price(buyer_age=age, price_t=BASE_PRICE)
    print(final_price)

start()

```

#### 30 - Fibonacci (Recursion)

Write a function fibonacci(n: int) -> int that returns the nth Fibonacci number. You may ignore negative arguments/

Test with values 5, 10, and 15.

You should get:

The 5th Fibonacci number is 5

The 10th Fibonacci number is 55

The 15th Fibonacci number is 610


```python
def fibonacci(n: int) -> int:
    a, b = 0, 1
    for i in range(n):
        a, b = b, a + b 
    return a


print(fibonacci(n=15))
print(fibonacci(n=5))
print(fibonacci(n=10))
```

#### 34 - Prime Number Checker

Write a function is_prime(n) that returns True if n is prime, otherwise False.

Try to get the following output:

Is 3 prime? -> True

Is 10 prime? -> False

Is 13 prime? -> True

```python
def is_prime(n:int) -> bool:
    n_dividers = 0
    for i in range(1, n+1):
        if n % i == 0:
            n_dividers += 1
    return n_dividers == 2

print(is_prime(n=13))
print(is_prime(n=10))
print(is_prime(n=3))
```
