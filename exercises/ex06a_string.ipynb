#### 7 - Function Conversion

Write a function convert_to_str(val) that prints both the original type and its string form. Test it with a number, range, tuple, list, set and dictionary.

```python
def convert_to_str(val):
    print(type(val))
    print(str(val))
dict = {
    '1': 'one',
    '2': 'Two',
    '3': 'Tree'
}

convert_to_str(val=100)
convert_to_str(val=range(10))
convert_to_str(val=(0, 2, 4))
convert_to_str(val=[4, 5, 6, 5])
convert_to_str(val={6, 6, 5, 6, 4, 2})
convert_to_str(val=dict)

```
