#### 3: File Reading
Write a program that asks the user for a filename and tries to open and read it. If the file doesn't exist, print a friendly error message. Ensure the file object is closed regardless of success or failure.   
Handle exceptions for:  
-   File not found.  
-   Permission errors.   
Respond with clear messages for each error type.



```python
file_name = input('File name: ')

try:
    with open(file=f'{file_name}.txt') as file:
        text = file.read()
except FileNotFoundError as error:
    print('File not found', error)
except PermissionError as  error:
    print('Permission error:', error)
else:
    print(text)
finally:
    print('Attempt to open file is done.')

```

#### 5: Binary Files
-   Write the following list of numbers as bytes to numbers.bin:   
data = [10, 20, 30, 40, 50]  
Then read it back and print the binary contents.   
-   Write the string "Hello, Binary!" to a binary file and then read it back, decoding it as UTF-8.

```python
data = [10, 20, 30, 40, 50] 

with open(file='numbers.bin', mode='wb') as file:
    file.write(bytes(data))

with open(file='numbers.bin', mode='rb') as file:
    text = file.read()
print(text)
print(list(text))

with open(file='numbers.bin', mode='ab') as file:
    file.write(bytes('\nHello, Binary!', encoding='UTF-8'))

with open(file='numbers.bin', mode='rb') as file:
    text_2 = file.read()

print(text_2.decode('utf-8'))

```

```python
text = 'Bonjour, Ã§a va ?'

with open(file='french.txt', mode='w', encoding='utf-8') as file:
    file.write(text)

with open(file='french.txt', encoding='utf-8') as file:
    print(file.read())


try:
    with open(file='french.txt', encoding='ascii') as file:
        print(file.read())
except UnicodeDecodeError as e:
    print('Error:', e)
```
