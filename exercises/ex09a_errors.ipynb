#### 1: Basic try-except
Write a program that asks the user for two numbers and divides the first by the second. Handle the case where the user enters zero as the second number.


```python
try:
    num_1 = float(input('Input any number: '))
    num_2 = float(input('Input any number: '))
    result = round(num_1 / num_2, 2) 
    print('Result = ', result)
except ZeroDivisionError as z:
    print(f'Error: {z} can not divide on 0')
except ValueError as v:
    print(f'Error: {v}  wrong value. Use numbers.')
```

#### 3: List Index Safety
Write a function safe_get(lst, index) that returns the element at the given index if it exists, or returns None if the index is out of range. Use exception handling to implement this.


```python
def safe_get(lst: list, index: int):
    try:
        return lst[index]
    except (IndexError, TypeError, SyntaxError):
        return None

try:
    print(safe_get(lst=[0, 2], index=3434))
except SyntaxError:
    print('Wont work')
    
```

#### 4: Dictionary Lookup
Write a program that stores student grades in a dictionary. Ask the user for a student name and display their grade. Handle the case where the student name doesn't exist in the dictionary.


```python
student_grades = {
    'Alex': 100,
    'Sim': 'So-SO',
    'Mariam': 'Ok'
}

def student_grade(dict_data: dict, dict_key):
    try:
        return dict_data[dict_key]
    except (KeyError, SyntaxError):
        print('Student with this name does not exist.')

student_grade(dict_data=student_grades, dict_key='Alex')
```

#### 5: Multiple Exception Types
Write a program that prompts the user to enter a password. Raise a `ValueError` if the password is less than 8 characters. Raise a `TypeError` if the password contains any spaces. Use try-except blocks to handle these errors and print appropriate messages.


```python
def password():
    password = input('Creat a password: ')
    if len(password) < 8:
        raise ValueError ('Password must be at least 8 characters')
    elif ' ' in password:
        raise TypeError ('Password must not contain spaces')
    return password

password()
```

#### 6: Multiple Exception Types
Write a program that tries to convert a userâ€™s input to an integer and then divides 100 by that number.   
-   Catch ValueError if the input is not a number
-   Catch ZeroDivisionError if the user enters 0


```python
def divicion():
    try:
        number = 100 / int(input('Input number: '))
    except ValueError as error:
        print(f'Error: {error}')
    except ZeroDivisionError as error:
        print(f'Error: {error}')
    else:
        return number

divicion()
```

#### 7: Age Validator with else and finally
Write a program that asks for a user's age and validates it. Use:

try to convert the input to an integer
except to handle invalid input
else to print a success message
finally to print "Validation complete"

Also check that the age is between 0 and 120.


```python
def age_validator():
    try:
        age = int(input('Your age: '))
        if not 0 <= age <=120:
            raise ValueError (f'Age must be between 0 and 120, got {age}')
    except ValueError as error:
        print(f'Error: {error}')
    else:
        print(f'Age {age} is valied')
        return age
    finally: 
        print('Validation is over.')

age_validator()
```

#### 9: Temperature Converter
Create a temperature converter with exception handling:

Ask the user for a temperature and scale (C or F)
Convert between Celsius and Fahrenheit
Handle invalid numeric input
Handle invalid scale input
Raise an exception if the temperature is below absolute zero (-273.15Â°C or -459.67Â°F)


```python
def t_converter():
    mode = input('C or F : ').lower()
    if mode not in ('c', 'f'):
        print('Wrong input. Must be C or F')
        return
    try:
        temp = int(input(f'Temp: {mode}'))
        if (mode == 'c' and temp < -273) or ( mode == 'f' and temp < -459):
            raise ValueError('Temp can not be less than -273.15 C or -459.67 F')
    except ValueError as error:
        print(f'Error: {error}')
    else:
        if mode == 'c':
            return  round(temp * 9 / 5 + 32)
        else:
            return round((temp - 32) * 5 / 9)
    finally:
        print('Programe is over')

t_converter()
```
